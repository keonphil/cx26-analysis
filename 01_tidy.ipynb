```python
import pandas as pd

def load_sheet(path, sheet, cols, names, construct, variant, experiment, source):
    df = pd.read_excel(path, sheet_name=sheet, usecols=cols, header=0)
    df.columns = names
    df["cell_id"] = [f"{source}_{i}" for i in range(1, len(df) + 1)]
    long = df.melt(id_vars="cell_id", var_name="condition", value_name="delta")
    long = long.dropna(subset=["delta"])
    long["pct_change"] = long["delta"] * 100
    long["construct"] = construct
    long["variant"] = variant
    long["experiment"] = experiment
    return long.drop(columns="delta")
```

```python
gfp = load_sheet("data/Delta.xlsx", "GFP (55 + Depol)", "A:B",
                 ["55", "depol"], "GFP", "WT", "response", "gfp_resp")
```

```python
gfp_ll = load_sheet("data/Delta.xlsx", "GFP LL", "A:B",
                 ["55", "depol"], "GFP", "LL", "response", "gfp_ll_resp")
```

```python
gfp_k125 = load_sheet("data/Delta.xlsx", "GFP K125", "A",
                 ["55"], "GFP", "K125", "response", "gfp_k125_resp")
```

```python
cpgfp_k125 = load_sheet("data/Delta.xlsx", "cpGFP K125", "A",
                 ["55"], "cpGFP", "K125", "response", "cpgfp_k125_resp")
```

```python
gfp_k125_2 = load_sheet("data/GFP_K125_2.xlsx", "Deltas", "A",
                        ["55"], "GFP", "K125", "response", "gfp_k125_resp2")
```

```python
tidy = pd.concat([gfp, cpgfp, gfp_dr, cpgfp_dr, gfp_ll, gfp_k125, gfp_k125_2, cpgfp_k125], ignore_index=True)
```

```python
tidy.groupby(["construct", "variant", "experiment", "condition"]).size()
```

```python
tidy.to_csv("data/tidy.csv", index=False)
```
